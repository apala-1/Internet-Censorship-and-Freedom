# Data Cleaning

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
freedom_score_data = pd.read_excel('data/freedom_Score.xlsx')
print(freedom_score_data.head())
```

```python
qog_basic_data = pd.read_csv('data/qog_basic_data.csv')
print(qog_basic_data.head())
```

```python
qog_imp_cols = qog_basic_data[['cname', 'year', 'wdi_internet', 'wdi_gdpcapcon2015', 'undp_hdi', 'vdem_libdem', 'ti_cpi']]
qog_cols_iso = qog_basic_data[['cname', 'ccodealp', 'year', 'wdi_internet', 'wdi_gdpcapcon2015', 'undp_hdi', 'vdem_libdem', 'ti_cpi']]
freedom_score_imp = freedom_score_data[['Country', 'Edition', 'Status', 'Total country score']]
print(qog_imp_cols.head())
print(freedom_score_imp.head())

print(qog_imp_cols.shape)
print(freedom_score_imp.shape)
```

```python
needed_years = [2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022]

new_qog = qog_imp_cols[qog_imp_cols['year'].isin(needed_years)]
new_qog_iso = qog_cols_iso[qog_cols_iso['year'].isin(needed_years)]
print(new_qog.head(20))
print(len(new_qog))
```

```python
print(new_qog.isna().sum())
print(freedom_score_imp.isna().sum())
```

```python
new_data = pd.merge(new_qog, freedom_score_imp, left_on=['cname', 'year'], right_on=['Country', 'Edition']).drop(['Country', 'Edition'], axis=1)
print(new_data.head())
new_data_iso = pd.merge(new_qog_iso, freedom_score_imp, left_on=['cname', 'year'], right_on=['Country', 'Edition']).drop(['Country', 'Edition'], axis=1)
```

```python
print(new_data.isnull().sum())

print(len(new_qog), len(new_data))
```

```python
new_data['wdi_internet'] = new_data['wdi_internet'].fillna(new_data['wdi_internet'].median())
new_data['ti_cpi'] = new_data['ti_cpi'].fillna(new_data['ti_cpi'].median())

new_data_iso['wdi_internet'] = new_data_iso['wdi_internet'].fillna(new_data_iso['wdi_internet'].median())
new_data_iso['ti_cpi'] = new_data_iso['ti_cpi'].fillna(new_data_iso['ti_cpi'].median())
```

```python
print(new_data.isna().sum())
```

```python
print(new_data.shape)
print(new_data.info())

print(new_data.describe())
print(new_data['Status'].value_counts())
```

```python
le = LabelEncoder()
new_data['status'] = le.fit_transform(new_data['Status'])
print(new_data['status'].value_counts())
new_data = new_data.drop(['Status'], axis=1)
print(new_data.head())

new_data_iso['status'] = le.fit_transform(new_data_iso['Status'])
new_data_iso = new_data_iso.drop(['Status'], axis=1)

# PF = 2
# F = 157
# NF = 147
```

```python
new_data['year'] = pd.to_datetime(new_data['year'], format='%Y')
new_data['status'] = new_data['status'].astype('category')
new_data.rename(columns={'cname':'country', 'Total country score':'freedom_score', 'status':'freedom_status'}, inplace=True)

new_data_iso['year'] = pd.to_datetime(new_data_iso['year'], format='%Y')
new_data_iso['status'] = new_data_iso['status'].astype('category')
new_data_iso.rename(columns={'cname':'country', 'Total country score':'freedom_score', 'status':'freedom_status'}, inplace=True)
```

```python
print(new_data.info())
```

```python
new_data.to_csv('cleaned_data/cleaned_data.csv', index=False)
new_data_iso.to_csv('cleaned_data/iso.csv', index=False)
```
